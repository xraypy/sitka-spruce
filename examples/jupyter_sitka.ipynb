This notebook will demonstrate using the Sitka Data Viewer from within a Jupyter Notebook.

Sitka can be used as an interactive viewer for HDF5 files and Zarr stores.  
This allows you to use the GUI to browse and explore the data, choosing which
datasets and arrays to work with.  

You can easily move data between the Jupyter session and the Sitka viewer so that
you can change what data arrays you use in downstream processing, and visualize 
and export results.


```python
# Step 0: if not already downloaded, get the Sitka example files
import os
from pathlib import Path
xrfmap1 = Path('xrf_map_seed.h5')
if not xrfmap1.exists():
    print("Downloading Sitka Examples")
    import get_sitka_examples

filelist = os.listdir('.')
print("Files: ", filelist)
```

```python
# Step 1: import and start the Sitka Viewer
# This will show the Sitka GUI in a separate window, 
# and read in all the data HDF5/Zarr files in the current folder.

from sitka_spruce import sitka_viewer
sview = sitka_viewer(folder='.')
```

```python
# The Sitka Viewer holds its data in the `data` attribute of the Sitka_Viewer.
# this a few useful attributes and methods
#   attribute/method    use
#    datasets           dict of {Filename: Data Object (h5py.File or zarr.Group)}
#    arrays             dict of {Name: ndarray} for named arrays
#    array_addrs        dict of {Name: access string} for how named arrays were created
#    add_dataset(filename, fileroot)    add a dataset into Sitka 
#    add_array(arrayname, ndarray)      add an array into the list of named arrays   
#    eval(string)                       execute a command in an Asteval interpreter.
#    export_hdf5(filename, list_of_arraynames)  export a list of named arrays to an HDF5 file.


print('Datasets: ', sview.data.datasets)
print('Named Arrays: ', sview.data.arrays)
print(help(sview.data.export_hdf5))

# datasets['Eiger9M_example.h5']['entry/data/data'][0,:,:]
```

```python
# if you copy an address with Ctrl-C in the Sitka window, you can copy it here and use it to extract data:

#  Ctrl-V gave ['xrf_map_seed.h5']['xrfmap/mca/counts'][:,:,250:350].sum(axis=2)
map1 = sview.data.datasets['xrf_map_seed.h5']['xrfmap/mca/counts'][:,:,250:350].sum(axis=2)

# now that you have one array, get another slice of the data
map2 = sview.data.datasets['xrf_map_seed.h5']['xrfmap/mca/counts'][:,:,350:450].sum(axis=2)

print(map1.shape, map1.dtype)

# now put the difference map into the Sitka session for viewing
sview.data.add_array('xrf_diff', map2-map1, "map2-map1")

# browse to "Named Arrays", click on 'xrf_diff' and then "Plot Current"
```
